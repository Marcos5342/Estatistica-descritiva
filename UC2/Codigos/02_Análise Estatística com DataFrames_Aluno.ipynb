Principais Estruturas de Dados do Pandas

• Series: É uma estrutura de dados unidimensional (semelhante a uma coluna de uma planilha ou a um array de uma dimensão) que armazena dados de qualquer tipo acompanhados de um rótulo de índice.  

• DataFrame: É uma estrutura de dados bidimensional tabular (semelhante a uma tabela de banco de dados ou planilha do Excel) composta por linhas e colunas rotuladas.


Principais Funções e Métodos

• read_csv(): Função usada para importar dados de um arquivo no formato CSV (Comma-Separated Values) e carregá-los na memória em formato de DataFrame. 

• head(n): Método que retorna as n primeiras linhas de um DataFrame ou Series (por padrão, exibe as 5 primeiras).  

• tail(n): Método que retorna as n últimas linhas de um DataFrame ou Series (por padrão, exibe as 5 últimas).  

• info(): Método que gera um resumo técnico do DataFrame, incluindo o total de linhas, nomes das colunas, quantidade de valores não nulos e o tipo de dado (dtype) de cada coluna. 

• shape: Atributo (não utiliza parênteses) que retorna uma tupla com as dimensões do DataFrame no formato (número_de_linhas, número_de_colunas).  

• Filtros Booleanos: Técnica que avalia uma condição lógica linha a linha (gerando uma Series temporária de valores True ou False) para selecionar apenas os registros que atendem ao critério desejado.

### Conceito Matemático
**Mediana**

Valor que ocupa a posição central dos dados ordenados. 
 
Exemplo

```python
lista = [10, 40, 20, 50, 30]
media = sum(lista) / len(lista)
print("A média é:", media)  # Resultado: 30.0
```

```python
import pandas as pd
import numpy as np

lista = [10, 40, 20, 50, 30]

# Opção 1: Usando Pandas (Series)
mediana_pandas = pd.Series(lista).median()
print(mediana_pandas)  # Resultado: 30.0

# Opção 2: Usando NumPy
mediana_numpy = np.median(lista)
print(mediana_numpy)  # Resultado: 30.0
```

Quartis

Quartis são valores que dividem um conjunto de dados ordenados em quatro partes iguais, onde cada parte representa exatamente 25% dos dados.  

Eles são fundamentais na estatística descritiva para entender a distribuição dos seus dados e identificar valores discrepantes (outliers).





📊 Os Três Quartis

• Q1 (Primeiro Quartil / Percentil 25): Deixa 25% dos dados abaixo dele e 75% acima. É o "meio do caminho" entre o menor valor e a mediana.  

• Q2 (Segundo Quartil / Percentil 50): É exatamente a Mediana. Divide a lista em duas partes de 50%.  

• Q3 (Terceiro Quartil / Percentil 75): Deixa 75% dos dados abaixo dele e 25% acima. É o "meio do caminho" entre a mediana e o maior valor.  

💡 IQR (Intervalo Interquartil): É a diferença entre o terceiro e o primeiro quartil (Q3 - Q1). Ele mostra onde estão concentrados os 50% centrais dos seus dados.


```python

import pandas as pd

# Criando uma série de dados de exemplo
serie_dados = pd.Series([10, 20, 30, 40, 50, 60, 70, 80])

# Calculando os quartis
q1 = serie_dados.quantile(0.25)
q2 = serie_dados.quantile(0.50) # Igual à mediana
q3 = serie_dados.quantile(0.75)

print(f"Q1 (25%): {q1}")  # Resultado: 27.5
print(f"Q2 (50%): {q2}")  # Resultado: 45.0 (Mediana)
print(f"Q3 (75%): {q3}")  # Resultado: 62.5
```

```python
import numpy as np

dados = np.array([
    12,15,17,20,22,
    25,28,30,35,40
])

q1 = np.percentile(dados,25)
q2 = np.percentile(dados,50)
q3 = np.percentile(dados,75)

print(f"Q1 = {q1}")
print(f"Q2 = {q2}")
print(f"Q3 = {q3}")
```

### loc, iloc e query


Para navegar e selecionar dados em um DataFrame, o Pandas oferece três ferramentas principais: loc, iloc e query(). Cada uma tem uma especialidade bem definida para filtrar linhas e colunas.




🗺️ A Diferença Prática: loc vs iloc

A regra de ouro para diferenciar os dois é: loc busca por rótulos (nomes) e iloc busca por posições numéricas (índices inteiros).

|Método	|Tipo de Busca	|Como funciona	|Exemplo Prático|
|-----|-----|-----|-----|
|loc	|Baseado em Rótulos	|Você passa o nome da linha e o nome da coluna.	|df.loc[0, 'Cidade']|
|iloc	|Baseado em Posições	|Você passa o número da linha e o número da coluna (estilo matriz, começando do 0).	|df.iloc[0, 2]|


📂 Exemplo usando a sua tabela de transações:

Imagine que queremos olhar apenas para a primeira linha do seu arquivo (índice 0).

```python
# Buscando a linha de índice 0 e a coluna chamada 'Cidade'
df.loc[0, 'Cidade']  # Retorno: 'Rio de Janeiro'
```

```python
df.loc[0:10,
       ["Cidade","Categoria","Valor_Total"]] # Retorna as linhas 0 a 10 e as colunas Cidade, Categoria e Valor_Total
```

```python
# Buscando a linha 0 e a coluna 2 (Data_Venda=1, Cidade=2)
df.iloc[0, 2]        # Retorno: 'Rio de Janeiro'
```

```python
df.iloc[0:10,1:4] # Retorna as linhas 0 a 9 e as colunas 1 a 3
```


✂️ Selecionando múltiplos dados (Slicing)  
 
• Com loc (Inclusivo): df.loc[0:2, 'Cidade':'Categoria'] vai trazer as linhas 0, 1 e 2 (o final é incluído).  
• Com iloc (Exclusivo): df.iloc[0:2, 2:4] vai trazer as linhas 0 e 1, e as colunas 2 e 3 (o final não é incluído, igual às listas do Python).  


🗣️ O Método query() (Filtros Legíveis)  

O query() permite fazer filtros booleanos (como os que vimos antes) escrevendo a condição como se fosse uma frase ou uma linha de código SQL, tornando a leitura muito mais limpa.  
Em vez de escrever uma sintaxe longa repetindo o nome do DataFrame:  

```python
df.query("Cidade == 'Rio de Janeiro'") # Retorna todas as linhas onde a cidade é 'Rio de Janeiro'
```

```python
df[(df['Cidade'] == 'Macaé') & (df['Quantidade'] > 5)] # Retorna todas as linhas onde a cidade é 'Macaé' e a quantidade é maior que 5
```

```python
df.query("Cidade == 'Rio de Janeiro' and Valor_Total > 10000") 
# Retorna todas as linhas onde a cidade é 'Rio de Janeiro' e o Valor_Total é maior que 10000
```

### Gráficos com Pandas

O Pandas facilita a criação de gráficos integrando-se diretamente com a biblioteca Matplotlib. Você pode gerar visualizações direto a partir de um DataFrame ou de uma Series apenas chamando o método .plot.  




1. Gráfico de Linha (Tendência)

É o padrão do Pandas. Excelente para mostrar a evolução de uma métrica ao longo do tempo.

```python
import pandas as pd
import matplotlib.pyplot as plt # Função da biblioteca matplotlib para plotagem de gráficos

# Convertendo a coluna para data e ordenando
df['Data_Venda'] = pd.to_datetime(df['Data_Venda']) # Converte a coluna 'Data_Venda' para o tipo datetime
df = df.sort_values('Data_Venda') # Ordena o DataFrame pela coluna 'Data_Venda'

# Gráfico da evolução do Valor Total das vendas
df.plot(x='Data_Venda', y='Valor_Total', kind='line', title='Evolução do Valor de Vendas') 
# mostrando a evolução do Valor_Total ao longo do tempo, com Data_Venda no eixo x e Valor_Total no eixo y, usando um gráfico de linha. 
  
plt.show() # apresenta o gráfico na tela
```


2. Gráfico de Barras (Comparações)

Ideal para comparar categorias, como o total vendido por cidade ou categoria de produto.

```python
vendas_cidade = (
    df.groupby("Cidade")
      ["Valor_Total"]
      .sum()
      .sort_values()
) # Agrupa os dados por cidade, soma o Valor_Total de cada cidade e ordena os resultados.

vendas_cidade.plot(
    kind="bar",
    figsize=(8,5),
    title="Faturamento por Cidade"
) # Cria um gráfico de barras com o faturamento por cidade
```

```python
# Agrupando os dados para somar o valor total por Cidade
vendas_por_cidade = df.groupby('Cidade')['Valor_Total'].sum()

# Criando o gráfico de barras
vendas_por_cidade.plot(kind='bar', color='skyblue', title='Total de Vendas por Cidade') # Cria um gráfico de barras com o faturamento por cidade
plt.ylabel('Valor Total (R$)') # Define o rótulo do eixo y
plt.xlabel('Cidade') # Define o rótulo do eixo x
plt.xticks(rotation=45) # Rotaciona os nomes das cidades para não embolar
plt.tight_layout()      # Ajusta o espaçamento
plt.show()
```


3. Histograma (Distribuição)

Mostra a frequência com que os valores aparecem. Perfeito para entender o comportamento dos preços ou valores totais.

```python
# Mostra a distribuição do valor total gasto nas transações
df['Valor_Total'].plot(kind='hist', bins=20, title='Distribuição do Valor Total das Transações') 
# Cria um gráfico de histograma com a distribuição do valor total das transações
plt.xlabel('Faixa de Valor (R$)') # Define o rótulo do eixo x
plt.ylabel('Frequência de Compras') # Define o rótulo do eixo y
plt.show()
```


4. Boxplot / Diagrama de Caixa (Visualizar os Quartis)

Lembra da nossa conversa sobre quartis e percentis? O Boxplot é o gráfico feito sob medida para isso. Ele desenha uma caixa que mostra visualmente o Q1, a Mediana (Q2), o Q3 e os valores discrepantes (outliers).

O Boxplot (ou Diagrama de Caixa) é uma ferramenta estatística visual utilizada para identificar a distribuição, a simetria, a dispersão e os valores discrepantes (outliers) de um conjunto de dados.

```python
# Mostra os quartis do preço unitário por Categoria de produto
df.boxplot(column='Preco_Unitario', by='Categoria') # Cria um boxplot para mostrar a distribuição do Preço Unitário por Categoria de produto
plt.title('Distribuição de Preços por Categoria') # Define o título do gráfico
plt.suptitle('') # Remove o título automático secundário
plt.ylabel('Preço Unitário (R$)') # 
plt.show()
```


🗺️ Anatomia do Boxplot (Como ler a imagem)

Cada uma das quatro "caixas" no seu gráfico representa uma categoria de produto. De baixo para cima, cada linha horizontal significa:
1. Linha Inferior ("Trilho" de baixo): O valor Mínimo dos preços (excluindo os outliers). Repare que todas as categorias possuem produtos bem baratinhos, quase encostando no 0.
2. Base da Caixa: O Primeiro Quartil (Q1 / Percentil 25). Significa que 25% dos produtos daquela categoria custam menos do que aquele valor (por volta de R$ 800 a R$ 1.000 no seu gráfico).
3. Linha Central (Dentro da caixa): A Mediana (Q2 / Percentil 50). É o coração do gráfico. Ela divide os produtos exatamente ao meio: 50% são mais baratos que essa linha e 50% são mais caros. Na categoria Móveis, a mediana está visivelmente um pouco mais alta (perto de R$ 1.800) do que em Eletrodomésticos (perto de R$ 1.700).
4. Topo da Caixa: O Terceiro Quartil (Q3 / Percentil 75). Indica que 75% dos produtos custam menos que esse valor (por volta de R$ 2.500 a R$ 2.600).
5. Linha Superior ("Trilho" de cima): O valor Máximo dos preços. Veja que o teto de preço de todas as categorias é muito parecido, batendo perto de R$ 3.500.


📦 A "Caixa" em si (IQR): A altura da caixa representa o Intervalo Interquartil (IQR = Q3 - Q1). Ela engloba os 50% centrais dos seus dados, ou seja, onde está a grande massa de vendas de produtos "padrão".



🎯 Quando e por que definir o uso do Boxplot?

O Boxplot é o gráfico ideal para três situações principais no seu dia a dia de análise de dados:  

• Comparar Categorias de forma justa: Em vez de comparar apenas a Média (que pode ser enganosa se houver um único produto extremamente caro), o Boxplot compara a estrutura inteira de preços. Pelo seu gráfico, descobrimos que a distribuição de preços entre as 4 categorias é muito equilibrada e semelhante, pois o tamanho das caixas e a altura das linhas são quase idênticos.  

• Visualizar a Dispersão (Variabilidade): Caixas mais compridas indicam dados mais espalhados e imprevisíveis. Caixas mais achatadas indicam dados mais concentrados e uniformes.  

• Caçar Outliers (Valores Discrepantes): Se houvesse algum produto custando R$ 10.000, por exemplo, o Pandas desenharia um "pontinho" isolado bem acima do limite de R$ 3.500. No seu gráfico atual, não existem outliers, o que significa que seus dados de preço estão bem contidos dentro da faixa normal esperada.



🛠️ O Guia do Argumento kind=  

O segredo do método .plot() está no argumento kind. Basta mudar essa palavra para alterar completamente o tipo de gráfico:  
• kind='line': Gráfico de linha (Padrão)  
• kind='bar': Barras verticais  
• kind='barh': Barras horizontais (ótimo para muitos nomes compridos)  
• kind='hist': Histograma  
• kind='box': Boxplot  
• kind='scatter': Gráfico de dispersão (requer passar colunas x e y para ver correlação)
