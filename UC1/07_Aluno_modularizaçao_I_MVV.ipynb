### 1. Conceito de Função 


#### Definição
Função é um bloco de código criado para realizar uma tarefa específica que pode ser reutilizada várias vezes.


Em Python, uma função com parâmetro é um bloco de código reutilizável que recebe valores externos (argumentos) para executar uma tarefa.

Conceito:  
- Função: bloco de código definido com def que pode ser chamado várias vezes.  
- Parâmetro: variável declarada na definição da função, usada para receber valores quando a função é chamada.  
- Argumento: valor passado para o parâmetro na chamada da função.  

#### Analogia  
Imagine um restaurante.  
Para preparar um prato existem etapas:  
•	Receber pedido  
•	Preparar prato  
•	Entregar pedido 


Cada etapa pode ser vista como uma função.

 #### Estrutura básica

``` 
def nome_da_funcao(parametro1, parametro2):  
    """
``` 

Python permite parâmetros nomeados (keyword arguments) e posicionais.

### 📚 Tipos de Parâmetros em Python

|Tipo de Parâmetro|	Sintaxe|	Descrição|	Exemplo|
|-----|-----|-----|-----|
|Posicional	|def func(a, b)	|Passados na ordem definida na função.	|func(1, 2)|
|Com valor padrão	|def func(a=10)	|Se não informado, usa o valor definido.	|func() → usa 10|
|Keyword (nomeado)	|def func(a, b)	|Passados com nome explícito.	|func(a=1, b=2)|
|Arbitrário Posicional	|def func(*args)	|Recebe múltiplos valores como tupla.	|func(1, 2, 3)|
|Arbitrário Nomeado|def func(**kwargs)	|Recebe múltiplos pares chave-valor como dict.	|func(x=1, y=2)|
|Somente Posicional	|def func(a, b, /)	|Argumentos antes de / só podem ser posicionais.	|func(1, 2)|
|Somente Nomeado|	def func(*, a, b)	|Argumentos após * só podem ser nomeados.	|func(a=1, b=2)|

````python
def exemplo(a, b=10, c=20, d=40, **kwargs): # kwargs é um dicionário que recebe argumentos nomeados adicionais
    print(a, b, c, d, kwargs)

exemplo( ) # Saída: TypeError: exemplo() missing 1 required positional argument: 'a'
exemplo(2)
exemplo(1, 5, c=15, d=30, extra=99)

````

#### Atividade 3 – Controle de Multa de Pesca

O limite permitido é de 100 kg.
Cada quilo excedente gera multa de R$ 40.
Criar uma função para calcular o valor da multa.



````python
def multa(p):
    if p - 100 >= 1:
        resultado = (p - 100) * 40
        print(f'A multa é de: R$ {resultado:.2f}.')
    else:
        return 'Não há multa até 100 Kg.'

multa(float(input("Digite um número.")))
````

#### Atividade 4 – Calculadora com função

Gere uma calculadora onde o usuário informa 2 números e realize as 4 funções matemáticas.  
Utilize funções para esa tarefa e informe que divisão por zero o resultado é zero.

````python
def calc(a,b):
    soma = a + b
    sub = a - b
    mult = a * b
    div = a / b
    print(f'A soma desses números é: {soma}. A subtração é: {sub}. A multiplicação é {mult}. E a divisão é: {div}.')
    if b == 0:
        print (f'A divisão por zero é igual a zero.')

calc(float(input('Insira o primeiro número.')),float(input('Insira o segundo número.')))
````

##### 5 - Calculadora IMC

Criar uma função que receba:  
•	Peso  
•	Altura  
Retornando o IMC e informar a classificação.


````python
def calc_imc(peso,altura):

    imc = peso / (altura **2)
    print(f'Você digitou peso de {peso} Kg e altura de {altura} metros.')
    print(f'Seu IMC é igual a {imc:.2f}.')

    if imc <= 0:
        print("Erro: números negativos ou igual a zero.")
    elif imc < 18.5:
        print("Segunda a OMS, seu IMC é classificado como magreza.")
    elif imc < 24.9:
        print("Segunda a OMS, seu IMC é classificado como normal.")
    elif imc < 29.9:
        print("Segunda a OMS, seu IMC é classificado como sobrepeso.")
    elif imc < 34.9:
        print("Segunda a OMS, seu IMC é classificado como obesidade nível 1.")
    elif imc < 39.9:
        print("Segunda a OMS, seu IMC é classificado como obesidade nível 2.")
    elif imc >= 40:
        print("Segunda a OMS, seu IMC é classificado como obesidade nível 3.")

calc_imc(float(input("Digite seu peso em quilogramas (se precisar expressar decimal, use ponto ao invés de vírgula).")),float(input("Agora digite sua altura em metros (novamente, se precisar expressar decimal, use ponto ao invés de vírgula).")))
````
