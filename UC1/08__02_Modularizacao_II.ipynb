#### Bibliotecas

Em Python, uma biblioteca é um conjunto de módulos relacionados, cada um contendo funções, classes e variáveis que resolvem problemas específicos. 

Podemos pensar nelas como caixas de ferramentas: ao importar uma biblioteca, você passa a ter acesso a recursos prontos sem precisar escrever tudo do zero 

Por exemplo, a biblioteca math oferece funções matemáticas básicas, enquanto NumPy permite operações matemáticas avançadas com arrays multidimensionais 


1. Vantagens de usar bibliotecas  
- Economia de tempo: evita escrever funções repetidas, acelerando o desenvolvimento. 

- Código mais organizado: cada biblioteca concentra funcionalidades específicas, facilitando a manutenção e leitura do código. 

- Menor propensão a erros: bibliotecas amplamente usadas são testadas por milhares de programadores, aumentando a confiabilidade. 

- Acesso a funcionalidades avançadas: como análise de dados (Pandas), visualização (Matplotlib), aprendizado de máquina (Scikit-learn) e inteligência artificial (TensorFlow) entre outras mais.



2. Tipos de bibliotecas  
- Biblioteca padrão: já vem instalada com Python, como os (sistema operacional), random (números aleatórios) e datetime (datas e horários).

- Bibliotecas externas: precisam ser instaladas via gerenciador de pacotes pip, como NumPy, Pandas, Matplotlib e Requests. 


3. Como usar  
Para utilizar uma biblioteca, basta importá-la no seu código com o comando import, por exemplo:

```python
import math  

print(math.sqrt(16))  # Saída: 4.0
```




#### Introdução à Biblioteca Random 

Função Random em Python  

A biblioteca random do Python é usada para gerar números pseudoaleatórios. Ela oferece diversas funções para trabalhar com números inteiros, flutuantes, sequências e distribuições estatísticas.

Fontes: <href>https://docs.python.org/pt-br/3/library/random.html 



Exemplo Básico: Gerar um Número Inteiro Aleatório
``` python
import random

# Gera um número inteiro aleatório entre 1 e 10 (inclusive)
numero = random.randint(1, 10)
print(f"Número gerado: {numero}")
```

Outras Funções Úteis  

1. random()  

    Gera um número flutuante aleatório entre 0.0 e 1.0.  
    ``` python  
    print(random.random()) # Exemplo de saída: 0.3744

    ```  

2. uniform(a, b)  

    Gera um número flutuante aleatório no intervalo [a, b].
    ``` python
    print(random.uniform(5.0, 10.0)) # Exemplo de saída: 7.3456

    ```

3. randrange(start, stop, step)

    Gera um número inteiro dentro de um intervalo com passos definidos.

    ``` python
    print(random.randrange(0, 10, 2)) # Exemplo de saída: 4 (pares entre 0 e 9)
    ```  

4. choice(seq)

    Seleciona aleatoriamente um elemento de uma sequência.

    ``` python
    cores = ['vermelho', 'azul', 'verde']
    print(random.choice(cores)) # Exemplo de saída: azul
    ```


5. shuffle(seq)

    Embaralha os elementos de uma lista.
    ``` python
    numeros = [1, 2, 3, 4, 5]
    random.shuffle(numeros)
    print(numeros) # Exemplo de saída: [3, 5, 1, 4, 2]
    ```


#### Considerações Importantes

Os números gerados pelo módulo random não são adequados para fins criptográficos. 
Para isso, use o módulo secrets.

Sempre que necessário reproduzir os mesmos resultados aleatórios (para testes), utilize random.seed(valor) para inicializar o gerador com uma semente fixa.

Essas funções tornam o módulo random extremamente versátil para aplicações como jogos, simulações e estatísticas!
