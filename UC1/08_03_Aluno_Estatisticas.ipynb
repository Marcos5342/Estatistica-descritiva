### Estatísticas Básicas

A estatística é a ciência e o ramo da matemática aplicada que estuda a coleta, organização, análise, interpretação e apresentação de dados. Ela funciona como uma caixa de ferramentas essencial para transformar números brutos em informações úteis, ajudando a identificar padrões, modelar incertezas e orientar a tomada de decisões em praticamente todas as áreas do conhecimento.  

Historicamente, o termo vem do latim status (Estado), pois suas primeiras aplicações na Antiguidade serviam para que os governantes pudessem realizar recenseamentos de habitantes e registrar nascimentos ou mortes.

**Os Dois Grandes Ramos da Estatística**

A estatística é dividida conceitualmente em duas frentes complementares:  
- Estatística Descritiva: Concentra-se em resumir e descrever as características de um conjunto de dados. Utiliza elementos visuais e matemáticos básicos como tabelas, gráficos, e medidas de tendência central (média, mediana e moda).  

- Estatística Inferencial (ou Indutiva): Utiliza teorias probabilísticas para extrair conclusões, estimar comportamentos ou fazer previsões sobre uma população inteira com base na análise de apenas uma amostra (um subconjunto representativo).

**Conceitos Fundamentais**

Para compreender o funcionamento das pesquisas estatísticas, existem quatro pilares fundamentais:


|Conceito|	Definição	|Exemplo Prático|
|-----|-----|-----|
|População	|Conjunto total de elementos que compartilham o objeto de estudo.	|Todos os eleitores cadastrados em um país.|
|Amostra	|Subconjunto menor extraído da população para viabilizar a análise.	|2.000 eleitores entrevistados em uma pesquisa eleitoral.|
|Variável	|A característica ou propriedade que se deseja medir em cada elemento.	|A intenção de voto, a idade ou a renda do eleitor.|
|Dados	|Os registros ou valores reais obtidos durante a medição da variável.	|"Candidato A", "34 anos", "2 salários mínimos".|


**Etapas do Método Estatístico**

Toda investigação estatística rigorosa segue um fluxo lógico de etapas:
1. Planejamento e Amostragem: Definição do problema, hipóteses e tamanho da amostra necessária.
2. Coleta de Dados: Obtenção das informações por meio de experimentos, censos ou questionários.
3. Crítica e Organização: Revisão dos dados brutos para eliminar erros e posterior tabulação.
4. Análise e Apresentação: Construção de gráficos informativos e aplicação de testes matemáticos.
5. Interpretação: Extração de conclusões lógicas dentro de margens de segurança e erro preestabelecidas.

### Tratamento de Erro:  
Em Python, try e except são usados para tratar erros sem interromper o programa abruptamente.

Estrutura básica


```python
try:
    # código que pode dar erro
    resultado = 10 / 0
except ZeroDivisionError:
    print("Não é possível dividir por zero.")
```

Se ocorrer um erro dentro do try, o Python procura um bloco except compatível.  
 No exemplo acima, como a divisão por zero gera ZeroDivisionError, a mensagem é exibida em vez de o programa travar.

Vários tipos de erro  
Você pode tratar erros diferentes separadamente:

```python
try:
    numero = int(input("Digite um número: "))
    divisao = 100 / numero
    print("Resultado da divisão:", divisao)
except ValueError:
    print("Você não digitou um número válido.")
except ZeroDivisionError:
    print("Não dá para dividir por zero.")
```

 else e finally  
Também existem else e finally:

```python
try:
    numero = int(input("Digite um número: "))
except ValueError:
    print("Entrada inválida.")
else:
    print("Número válido:", numero)
finally:
    print("Esse bloco sempre executa.")
```

- else: roda só se não houver erro no try
- finally: roda sempre, com erro ou sem erro

Capturando qualquer erro  
Você também pode usar:

```python
try:
    x = 1 / 0
except Exception as e:
    print("Ocorreu um erro:", e)
```

Mas o ideal é capturar erros específicos, porque isso deixa o código mais seguro e fácil de manter.

```python
cardapio = {
    1: ("Sushi", 35),
    2: ("Temaki", 28),
    3: ("Yakisoba", 30),
    4: ("Sashimi", 40)
}
bebidas = {
    1: ("Refrigerante", 8),
    2: ("Suco", 7),
    3: ("Água", 4)
}

sobremesas = {
    1: ("Pudim", 12),
    2: ("Sorvete", 10),
    3: ("Mousse", 9)
}

def escolher_item(titulo, opcoes, obrigatorio=True):
    print(f"\n=== {titulo} ===")
    for codigo, (nome, preco) in opcoes.items():
        print(f"{codigo} - {nome} - R${preco:.2f}")
    if not obrigatorio:
        print("0 - Não desejo")

    while True:
        try:
            escolha = int(input("Escolha uma opção: "))
            if not obrigatorio and escolha == 0:
                return None
            if escolha in opcoes:
                nome, preco = opcoes[escolha]
                return nome, preco
            print("Opção inválida.")
        except ValueError:
            print("Digite um número válido.")

pedido = [
    ("Prato", escolher_item("Pratos", cardapio))
]

bebida = escolher_item("Bebidas", bebidas, obrigatorio=False)
if bebida:
    pedido.append(("Bebida", bebida))

sobremesa = escolher_item("Sobremesas", sobremesas, obrigatorio=False)
if sobremesa:
    pedido.append(("Sobremesa", sobremesa))

print("\n=== Confirme seu pedido ===")
for categoria, (nome, preco) in pedido:
    print(f"{categoria}: {nome} - R${preco:.2f}")

confirmacao = input("Confirmar pedido? (s/n): ").strip().lower()

if confirmacao == "s":
    total = sum(preco for _, (_, preco) in pedido)
    print(f"\nTotal da conta: R${total:.2f}")
else:
    print("Pedido cancelado.")
def mostrar_cardapio():
    print("=== Cardápio ===")
    for codigo, (prato, preco) in cardapio.items():
        print(f"{codigo} - {prato} - R${preco:.2f}")

def realizar_pedido():
    try:
        pedido = int(input("Escolha o número do prato: "))
        if pedido in cardapio:
            return pedido
        print("Prato inválido.")
    except ValueError:
        print("Digite um número válido.")
    return None

def fechar_conta(pedido):
    prato, preco = cardapio[pedido]
    print(f"Pedido: {prato}")
    print(f"Total da conta: R${preco:.2f}")



```
